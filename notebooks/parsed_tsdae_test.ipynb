```python
# Importy
import pandas as pd
from sentence_transformers import SentenceTransformer, losses, models, InputExample
from torch.utils.data import DataLoader
import os
```

```python
# Wczytanie danych IFC
df = pd.read_csv("../data/parsed_tsdae_dataset.csv")
texts = list(set(df["Text"].dropna().astype(str).tolist()))
print(f"{len(texts)} unikalnych opisów")
```

```python
# Przygotowanie danych do TSDAE
train_samples = [InputExample(texts=[text, text]) for text in texts]
train_dataloader = DataLoader(train_samples, batch_size=16, shuffle=True)
```

```python
# Model
model = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")
train_loss = losses.MultipleNegativesRankingLoss(model)
```

```python
# Trening modelu
output_path = "../models/tsdae_model_parsed_ifc"
model.fit(
train_objectives=[(train_dataloader, train_loss)],
epochs=1,
warmup_steps=100,
output_path=output_path,
show_progress_bar=True
)
```
