```python
# Importy
import pandas as pd
from sentence_transformers import SentenceTransformer, InputExample, losses, models
from torch.utils.data import DataLoader
```

```python
# Dane do few-shot fine-tuning
FILE_PATHS = [
"../data/fewshot_dataset/file01.csv",
"../data/fewshot_dataset/file02.csv"
]
```

```python
# Datset do few-shot fine-tuning
examples = []

for path in FILE_PATHS:
    df = pd.read_csv(path)
    for i, row in df.iterrows():
        if pd.notna(row["Text"]) and pd.notna(row["ClassificationCode"]):
            examples.append(InputExample(texts=[row["Text"], row["ClassificationCode"]]))

print(f"Załadowano {len(examples)} przykładów.")
```

```python
# Dataloader
train_dataloader = DataLoader(examples, shuffle=True, batch_size=8)
```

```python
# Model (po TSDAE)
model_path = "../models/parsed_tsdae_model" # lub inna ścieżka, jeśli inny folder
model = SentenceTransformer(model_path)
```

```python
# Funkcja straty (loss)
loss = losses.MultipleNegativesRankingLoss(model)
```

```python
# Trening
model.fit(
train_objectives=[(train_dataloader, loss)],
epochs=3,
warmup_steps=10,
show_progress_bar=True,
output_path="../models/fewshot_finetuned"
)

print("Trening zakończony. Model zapisany do ../models/fewshot_finetuned")
```
